### 1.1.3. Package managers

#### Conda  
- **Conda** is both a package manager and an environment manager.
- This means that it can install libraries, but also create and manage isolated environments with specific versions of Python and dependencies.  
- It supports not only Python packages but also packages written in other languages such as C or R, which makes it more versatile for data science and machine learning.  
- Conda installs **precompiled binaries**, so users do not need to compile code during installation, which reduces errors and simplifies setup.  
- Example commands:  
  ```bash
  conda create -n myenv python=3.10
  conda activate myenv
  conda install numpy pandas matplotlib
  conda list

#### Pip  
- **Pip** is the default package manager that comes with Python.
- It installs Python packages directly from the **Python Package Index (PyPI)**.
- An important advantage of pip is that **packages on PyPI are often more up to date than in conda repositories**, meaning that the newest versions of libraries are usually available first through pip. This makes it the preferred choice when developers need the latest features of a package. 
- Unlike conda, pip only manages Python packages and does not handle system-level dependencies, which sometimes leads to compatibility issues.  
- Pip installs **source distributions** by default, which may require local compilation. This can cause errors if compilers or system libraries are missing.  
- Pip does not natively create or manage environments, so it is usually combined with **venv** or **virtualenv** to isolate projects.  
- Example commands:  
  ```bash
  python -m venv myenv
  source myenv/bin/activate   # Linux/Mac
  myenv\Scripts\activate      # Windows
  pip install numpy pandas matplotlib
  pip list


#### Common practice

- A common practice is to combine both:
    - Create the environment with conda.
    - Install core scientific packages with conda (e.g., NumPy, TensorFlow, PyTorch).
    - Use pip to install additional Python libraries not available on conda.

### 1.1.5.	Jupyter notebooks and Jupyter lab

#### Jupyter Notebooks 

- **Jupyter Notebooks** are interactive documents that allow you to combine executable Python code, formatted text using Markdown, mathematical equations written in LaTeX, and visualizations all in the same environment. This makes them extremely useful for experimentation, teaching, and documenting data analysis or deep learning workflows.  

- Notebooks provide a **cell-based structure**, where each cell can contain either code or text. You can run cells individually, making it easy to test small pieces of code, visualize intermediate results, and debug step by step without rerunning the entire script.  
- They are widely used in deep learning because you can **train models, plot graphs, and inspect data** interactively, all within the same document. Notebooks also support rich media, including images, videos, and interactive widgets, which help in explaining results or building tutorials.  


#### Jupyter Lab

- **JupyterLab** is the next-generation interface of Jupyter Notebooks, designed to provide a more **flexible and powerful environment**. It allows you to open multiple notebooks, terminals, text files, and consoles in a single workspace, similar to a full integrated development environment (IDE).  

- JupyterLab supports **extensions and plugins**, enabling features such as Git integration, interactive debugging, real-time collaboration, and visual widgets for data exploration. This makes it particularly useful for **large projects or research workflows** where multiple tools and notebooks need to be managed simultaneously.  

- Both Jupyter Notebooks and JupyterLab are **cross-platform and browser-based**, so they can be used on local machines, remote servers, or cloud platforms like Google Colab, providing flexibility for experimentation and collaboration.  

- Overall, **Jupyter Notebooks are ideal for interactive exploration and prototyping**, while **JupyterLab** adds organization and advanced functionality, **making it suitable for more complex projects or team-based development**.

#### Unification

- Starting from **Jupyter Notebook version 7**, the architecture of Notebook has been **unified with JupyterLab**. This means that when you run `jupyter notebook`, you are actually running a simplified version of JupyterLab behind the scenes.  
- Notebook 7 maintains the classic Notebook interface but uses the **modern Lab architecture**, allowing access to extensions, multiple file handling, and improved terminal integration.  
- In other words, Notebook 7 behaves like a Notebook “wrapped” in JupyterLab, reducing the differences between the two environments while keeping the familiar interface for users.  

## 1.2 Training using GPUs and TPUs  

- **Deep learning models require significant computational resources** because they involve training with large datasets and millions of parameters. Standard CPUs can be too slow for this process, so researchers and developers usually rely on **GPUs (Graphics Processing Units)** and **TPUs (Tensor Processing Units)** to accelerate training.  
- **GPUs** are highly parallel processors originally designed for graphics rendering, but their architecture is well suited for matrix operations and deep learning computations.  
- **TPUs**, developed by Google, are specialized hardware accelerators specifically designed for tensor operations in machine learning. They are particularly efficient for training and inference with TensorFlow models.  

### 1.2.1 Local GPUs  

- Local GPUs refer to graphics cards installed on a personal computer or workstation. Popular options include **NVIDIA GPUs** with CUDA support, which are widely used in deep learning.  
- Training on local GPUs allows for **full control of the environment**, installed packages, and data, making it a good option for small to medium-scale projects.  
- However, local GPUs can be **expensive to acquire and maintain**, and upgrading hardware frequently is often necessary to keep up with the latest deep learning requirements.  

### 1.2.2 Google Colab GPUs  

- **Google Colab** offers free access to GPUs through a cloud-based Jupyter notebook environment. It provides **Tesla T4, P100, or V100 GPUs** depending on availability.  
- This option is ideal for **students, researchers, or small projects**, since it eliminates the need for expensive hardware.  
- Limitations include **session timeouts**, restricted GPU availability, and limited RAM. For longer training or more demanding projects, Google Colab Pro or Pro+ subscriptions offer better resources.  
- Colab also supports **TPUs**, making it possible to experiment with specialized hardware without cost.  

### 1.2.3 Supercomputing Center GPUs  

- Supercomputing centers provide access to **large-scale GPU clusters** designed for high-performance computing (HPC). These clusters often include thousands of GPUs interconnected with high-speed networks.  
- They are ideal for **large-scale experiments, massive datasets, or complex deep learning models** that cannot be trained efficiently on local or cloud-based free resources.  
- Access is usually granted through **research institutions, grants, or paid services**, and often requires familiarity with job schedulers (e.g., SLURM) and command-line tools.  
- Supercomputing centers allow **parallel training across multiple GPUs or nodes**, significantly reducing training time for very large models.

### 1.3.3. MNIST: Preparing the data

* We need to prepare the data to what the network expects from them.
* The images are an array of dimensions `(60000, 28, 28)` of type `uint8` with values of `[0, 255]`.
* First, they must be transformed into an array of dimensions `(60000, 28 * 28)` 
  - The image is flatten to a simple array of `28 * 28 = 784` values.
  - We use the NumPy `reshape` function that changes the shape of an array without changing its data (https://numpy.org/doc/stable/reference/generated/numpy.reshape.html)


````python
train_images = train_images.reshape((60000, 28 * 28))
test_images = test_images.reshape((10000, 28 * 28))
````

* Second, they must be tranformed to a float type `float32` with values between 0 and 1.
    - Each value ranges from 0 to 255 (levels of gray) and should be normalized to values between 0 and 1 (dividing by 255).
    - We use the NumPy `astype` function that copies the array and cast its values to a specified type.
    - When we divide an `ndarray` by a number we divide each element of the array by that number.

````python
train_images = train_images.astype('float32') / 255
test_images = test_images.astype('float32') / 255
````

* We also need to categorically encode the labels using the *one-hot* encoding.
* In this encoding we convert an integer into an array of integers in which all values are set to zero except the value that corresponded to the original integer which is set to one.
* We use the `to_categorical` function that converts integers to a binary class matrix (https://keras.io/api/utils/python_utils/)

````python
train_labels[2]
````

````python
from keras.utils import to_categorical
train_labels = to_categorical(train_labels)
test_labels = to_categorical(test_labels)
````

### 1.3.4. MNIST: Building the Model and Layers

#### Sequential model

* We are going to use the Keras sequential model.
* We use the `Sequential` class (https://keras.io/api/models/sequential/) to create a sequential model.

````python
import keras
from keras import layers

model = keras.Sequential()
````

#### Input layer

* The first layer to create is an input layer that receives the input (https://keras.io/api/layers/core_layers/input/).
* The most important parameter in the constructor of `Input` is `shape`, a tuple of integer that indicates the shape of the input.
* For instance, `shape=(784,)` indicates that the expected input will be batches of 784-dimensional vectors (our 28x28 images). 
* Elements of this tuple can be `None` representing dimensions where the shape is not known.
* We add this layer to the model using the `add` method.

````python
model.add(layers.Input(shape=(784, )))
````

#### Artificial Neurons

* We can remenber that an **artificial neuron** is composed of the following elements:
    * **Input values**. Which are supplied to the neuron.
    * **Weights**. Associated to each input value.
    * **Sum function**. That sums the values of the inputs multiplied by their respective weights (dot product).
    * **Bias**. Numerical value used to adjust the result of multiplying the inputs by the weights to obtain better predictions.
    * **Activation function**. Function that converts neuron input into neuron output.

#### Dense layers

* Dense layers are layers formed by artificial neurons each one is fully connected with the previous and subsequent layers.
* In Keras `Dense` (https://keras.io/api/layers/core_layers/dense/) represents a densely-connected neural network layer that implements the operation `output = activation(dot(input, kernel) + bias)`.
* The typical parameters for the `Dense` constructor are:
    - units: the number of neurons in the layer.
    - activation: the activation function used in each neuron.
    - input_shape: when passed Keras will create an input layer to insert before the current layer. This can be treated equivalent ot explicitly defining an `Input` Layer.
* In this case we are going to create a dense layer that receives the 784 inputs and produces 512 outputs activated by the ReLU function.

````python
model.add(layers.Dense(512, activation="relu"))
````

#### ReLU (rectified linear unit) activation function

<!-- https://www.v7labs.com/blog/neural-networks-activation-functions -->

* The ReLU function is defined as: $f(x) = \max(0, x)$
* It is a **popular activation function in *deep learning*** because:
    - It is computationally efficient as it is easy to compute and only a certain number of neurons are activated.
    - Large values of $x$ correspond to large values of ReLU, so the function does not saturate as occurs with other alternatives such as the sigmoid function or the hyperbolic tangent. 
    - Therefore ReLU accelerates the convergence of gradient descent towards the global minimum of the loss function.
* The drawback is the **Dying ReLU problem**:
    - The negative side of the graph makes the gradient value zero. 
    - Due to this reason, during the backpropagation process, the weights and biases for some neurons are not updated. This can create dead neurons which never get activated. It decreases the model’s ability to fit or train from the data properly. 
    - Some modifications are considered to solve this problem such as the Leaky ReLU or the parametric ReLU.
    

#### Output layer

* The output layer is also a dense layer but in this case with 10 neurons, each of which will represent a digit.
* The output value of each neuron of the output layer will indicate the probability that the current digit corresponds to the digit represented by that neuron.
* The total probability must add up to one.
* To obtain these values we use the activation function *softmax*.

````python
model.add(layers.Dense(10, activation="softmax"))
````

#### Softmax activation function

* The softmax function or normalized exponential function converts a vector of $K$ real numbers into a probability distribution of $K$ possible outcomes.
* The softmax function is often used as the last activation function of a neural network to normalize the output of a network to a probability distribution over predicted output classes.
* The standard softmax function $\sigma : \mathbb{R}^K \to (0, 1)^K$ is defined when $K \ge 1$ by the formula:



$\sigma(\mathbf{z})_i = \frac{e^{z_i}}{\sum_{j=1}^K e^{z_j}} \ \ \text{ for } i = 1, \dotsc, K \text{ and } \mathbf{z} = (z_1, \dotsc, z_K) \in \mathbb{R}^K.$

### 1.3.5. MNIST: Learning process

#### Defining the learning parameters

* Before training a model, it is necessary to set up the learning process, which is done through the `compile` method. 
* We need to pick three more things as part of the compilation step:
    - A **loss function**: How the model will be able to measure its performance on the training data, and thus how it will be able to steer itself in the right direction.
    - An **optimizer**: The mechanism through which the model will update itself based on the training data it sees, so as to improve its performance.
    - **Metrics** to monitor during training and testing. Here, we’ll only care about accuracy (the fraction of the images that were correctly classified).

````python
model.compile(optimizer="adam", 
              loss="categorical_crossentropy",
              metrics=["accuracy"])
````

#### Training

* We’re now ready to train the model, which in Keras is done via a call to the model’s `fit()` method—we fit the model to its training data.

* The main parameters of the `fit()` method are (https://keras.io/api/models/model_training_apis/):
    - The **input data**: the train images.
    - The **target data**: the train labels.
    - The **epochs**: Number of iterations over the entire input and target data. 
        * Using the input data several times we increase the accuracy of our model. 
        * But we have to take care of not causing overfitting.
    - The **batch_size**: Number of samples per gradient update.
        * With a batch size of 1 we upgrade the network weights after each sample has passed throught the network, this is slow and consume a lot of resources.
        * We can group the input data in batches and only update the weights after entering in the network all the samples of the batch.
        * The learning consume less resources, but since we are propagating backwards the average error of all the batch samples, the quality of the model may degrade and may ultimately be unable to generalize well on data it hasn't seen before.

* Two quantities are displayed during training: the loss of the model over the training data, and the accuracy of the model over the training data.

````python
model.fit(train_images, train_labels, epochs=5, batch_size=64)
````
